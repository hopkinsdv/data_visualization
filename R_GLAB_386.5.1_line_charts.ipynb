```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.arange(1, 12, 2)
y = np.arange(2, 13, 2)

# Create a line plot with a dashed linestyle
plt.plot(x, y, linestyle='-.', marker='o')
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot(x, y, linestyle="-.", marker="*")
plt.show()
```

```python
plt.plot(x, y, linestyle=":")
plt.show()
```

```python
x = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
```

```python
y = np.array([3, 8, 1, 10, 4, 200, 45, 67, 56, 98, 100])
```

```python
# Create a dashed line, with diamond markers
plt.plot(y, linestyle="--", marker="D")
plt.show()
```

```python
# Using the opposite approach, linestyle of dashdot, and square markers
fig, ax = plt.subplots()

ax.plot(y, linestyle="-.", marker="s")
plt.show()
```

```python
# diamond marker, red marker color, marker width
plt.plot(y, marker='D' , markerfacecolor='red' , markersize=7, markeredgewidth=2)
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 170, 190, 160, 300, 310, 320, 330])
```

```python
# Create a lineplot, custom cuolor, dotted linestyle, diamond markers
plt.plot(x, y, linestyle=":", marker="D", color=(0.5, 0.5, 0.8))
plt.plot(x, [170, 200, 160, 225, 275, 300, 375, 400, 300, 125])
plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")
plt.show()
```

```python
# Create a lineplot, custom cuolor, dotted linestyle, diamond markers
plt.plot(x, y, linestyle=":", marker="D", color="#FF5733")
plt.title("Sports Watch Data", color="blue")
plt.xlabel("Average Pulse", color="green")
plt.ylabel("Calorie Burnage", color="red")
plt.show()
```

```python
years = [1970, 1980, 1985, 1990, 1995, 2000, 2005, 2010, 2015, 2020, 2025, 2030, 2035]
pops = [3.6, 4.0, 4.4, 4.8, 5.3, 5.7, 6.1, 6.5, 7.3, 7.3, 7.3, 7.4, 7.5]
```

```python
# title of 'World Population' -  blue
# labels growth by year, population in billions,  green red
plt.plot(years, pops, linestyle="--", marker="*", markerfacecolor="orange")
plt.title("World Population", color="blue")
plt.xlabel("Growth by Year", color="green")
plt.ylabel("Population (billions)", color="red")
plt.show()
```
